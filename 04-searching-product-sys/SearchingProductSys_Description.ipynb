# Проект модуля. Система поиска подходящих товаров на основе дообученной CLIP модели

Представьте что вы работаете в крупном онлайн-магазине одежды. Вам нужно реализовать такую систему: пользователь описывает вещь словами, система находит и показывает наиболее близкие позиции из базы. Компания выходит на международный рынок, поэтому система будет работать на английском языке.

## Краткое описание

### План проекта
В этом проекте вам нужно:
1. дообучить модель CLIP на датасете товаров из магазинов одежды,
2. реализовать систему поиска изображений товаров по текстовым запросам.

### Датасет

<a href="https://www.kaggle.com/datasets/nirmalsankalana/fashion-product-text-images-dataset">Скачайте датасет с kaggle.com</a>. Fashion Product Images and Text Dataset.

Датасет содержит картинки и таблицу `data.csv`. В этой таблице такие поля:
- `image` — название файла изображения в папке с изображениями.
- `description` — описание элемента одежды, изображённого на картинке.
- `display name` — название элемента на картинке.
- `category` — категория элемента одежды.

В проекте вам понадобятся только первые два столбца: image — для визуальных данных, description — для текстовых.

## Задание

Выполните проект по этапам.

### Этап 1. Fine-tuning CLIP


1. **Работа с данными**

    - Загрузите датасет и визуализируйте часть данных. Посмотрите разрешение картинок. Возьмите несколько случайных картинок и проанализируйте, насколько текстовые описания подходят к картинкам.
    - Предобработайте данные: уберите дубликаты, уберите пустые данные из таблицы, разделите данные на трейн и тест (10% от всех данных).
    - Реализуйте свой класс датасета, который будет подгружать картинки и соответствующие им текстовые описания.

2. **Работа с моделью**

    - Загрузите предобученную модель CLIP из библиотеки `transformers`. Можете использовать `openai/clip-vit-base-patch32` или другую.
    - Прогоните через модель несколько примеров из датасета и посмотрите, какие получаются CLIP-скоры.

3. **Дообучение**

    - Реализуйте цикл обучения модели CLIP со стандартной функцией потерь. Во время обучения логируйте среднюю функцию потерь и CLIP score для каждого батча.
    - Реализуйте функцию валидации, которая будет замерять средний CLIP score на всей тестовой выборке в конце эпохи обучения.
    - Дообучите модель. Кривая функции потерь должна сойтись через 3-4 эпохи обучения.
    - Сделайте checkpoint-ы модели. Можно делать их в конце каждой эпохи или чаще.
    - После обучения выведите кривые функции потерь и CLIP score на трейне и CLIP score на валидации. **Целевые значения среднего CLIP score должны быть выше 30**.


**Результат:** описаны результаты анализа данных (в свободной форме), написаны функции трейна и валидации модели, получен чекпоинт дообученной модели и выведены кривые обучения и валидации.

### Этап 2. Реализация системы поиска товаров

1. **Работа с данными**

Возьмите весь датасет целиком. Чтобы система поиска товаров работала быстро, нужно заранее посчитать векторные представления всех картинок. Загрузите чекпоинт дообученной модели и сохраните все векторные представления изображений.

2. **Функция поиска товаров**
- Напишите функцию поиска товаров.

Функция должна принимать такие аргументы:
- модель,
- данные из всего датасета,
- текстовый запрос пользователя,
- количество наиболее релевантных товаров, которые будут показываться пользователю.

Формат аргументов выберите самостоятельно. Если нужно, передавайте в функцию дополнительные аргументы.

Функция должна вычислять векторное представление текстового запроса пользователя, находить заданное количество наиболее актуальных картинок с товарами и возвращать их.

3. Протестируйте написанную функцию. Введите несколько запросов, например: "red skirt", "blue sunglasses", "mickey mouse". Отрисуйте изображения товаров, которые выводит функция.

**Результат:** реализована поисковая система, способная принимать текстовые запросы и возвращать наиболее релевантные изображения из датасета. Система должна работать эффективно, то есть не пересчитывать каждый раз векторные представления картинок.
